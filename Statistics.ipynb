## Typer av statisik 

Decriptive statistics: beskriver och sammanfattar data 

Inferential statistics: använder stickprov för att inferera om en större population. Tex hur stor procent av människor kör bil till jobbet? (baserat på vårt stickprov) 


## Typer av data  

### Numerisk: 

#### Continuous: 
hastighet eller tid, kvantiteter som kan mätas 

#### Discrete: 
count , som antal husdjur eller paket som skickats  

### Kategorisk:  

Om en variabel som är av siffror, inte kan ordnas, då måste den vara kategorisk, som postnummer. 

#### Nominal 
är oordnat, som gift/ogift, eller land. Kan representeras som 1/0. 

#### Ordinal data 
har en särskild inneboende ordning, tex i en enkät när man ska svara 1-5 hur mkt man håller med
