### Context manager

- Sets up a context
- Runs your code
- Removes the context

Analogy: caterers that set up tables with food and drinks. They let you and your friends have a party. Then they clean up and remove the tables.

the open() function is a context manager. It:
- sets up a context by opening a file
- lets you run any code you want on that file
- removes the context by closing the file

The key 'with' lets Python know that you are trying to enter a context. A context manager can take arguments like any other function. You end the 'with' statement with a colon as if you were writing a for loop or an if statement. They are compound statements, having indented statements. Adter the indentation, the cleaning happens. Some context managers want to return a value that you can use inside the context. By adding 'as' and a variable name, you can assign the returned value to the variable name.


    with <context-manager>(<args>) as <variable-name>:
        run your code here

Ex:

    with open('my_file.txt') as my_file:
        text = my_file.read()
        length = len(text)
    print ('The file is {} characters long'.format(length))

By adding 'as my_file' to the with statement, we assign the file to the variable 'my_file'

5. add the @contextlib.contectmanager decorator 

### The 'yield' keyword

When you write this word, it mean that you are going to return a value, but you expect to finish the rest of the function at some point in the future. 

The value that your context manager yields can be assigned to a variable in the with statement by adding 'as' variable-name.

A context manager function is technically a generator that yields a single value.

The ability for a function to yield control and know that it will get to finish running later is what makes context managers so useful.

Set up code: before yielding

Teardown code: after yielding

This setup/teardown behaviour allows a context manager to hide things like connecting and disconnecting from a database so that a context manager can just perform operations on the database without worrying about the underlying details.

    @contextlib.contextmanager
    def database(url):

    db = postgres.connect(url)

    yield db

    db.disconnect()


    url = ''
    with database(url) as my_db:
    course_list = my_db.execute(
        'SELECT * FROM courses'
    )

This context manager yields a specific value, the database connection - that can be used in the context block.

Some context managers don't yield an explicit value.

## Closures

is a tuple of variables that are no longer in scope, but that a function needs in order to run.

Python's way of attaching nonlocal variables to a returned function.

When one function returns another function, Python helpfully attaches any non-local variables that the second function is going to need to the function object. Those variables get stored in a tuple in the double underscore closure attribute of the function. 

You can see the value of the variable by accessing the cell_contents of the item

    func.__closure__[0].cell_contents

 you can modify, delete, or overwrite the values needed by the nested function, but the nested function can still access those values because they are stored safely in the function's closure
