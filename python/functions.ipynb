## Nested functions

For example if you want to perform a computation several times in a function.

## Returning functions

```python
def raise_val(n):
    """Return the inner function"""

    def inner(x):
        """Raise x to the power of n"""
        raised = x ** n
        return raised
    
    return inner

square = raise_val(2)
cube = raise_val(3)

print(square(2), cube(4))
```

### Flexible arguments: *args

The user can put as many arguments as they want!


```python
def add_all(*args):
    """Sum all values in *args together"""

    #Initialize sum
    sum_all = 0

    #Accumulate the sum
    for num in args:
        sum_all += num 

    return sum_all

add_all(1, 2, 3)
```

### Flexible arguments: **kwargs

keyword arguments - arguments preceded by identifers

```python
def print_all(**kwargs):
    """Print out key-value pairs in **kwargs."""

    #Print out the key-value pairs
    for key, value in kwargs.items():
        print(key + ": " + value)

print_all(name='dumbledore', job='headmaster')

```

## Lambda functions

quick way to write a function on the fly

### Anonymous functions

when we give a lambda to map()

Function map() takes two arguments map(func, seq)
map() apples the function to ALL elements in the sequence

We call it an anonymous function, because we don't even have to name the lambda function when we pass it to map()

```python
raise_to_power = lambda x, y: x ** y

raise_to_power(2, 3)
```

```python
nums = [48, 6, 9, 21, 1]

square_all = map(lambda num: num ** 2, nums)

print(square_all)
```

```python
print(list(square_all))
```

## Filter() and lambda functions

filter() is a function that offers a way to filter out elements from a list that don't satisdy a certain criteria

```python
# Create a list of strings: fellowship
fellowship = ['frodo', 'samwise', 'merry', 'pippin', 'aragorn', 'boromir', 'legolas', 'gimli', 'gandalf']

# Use filter() to apply a lambda function over fellowship: result
result = filter(lambda member: len(member) > 6, fellowship)

# Convert result to a list: result_list
result_list = list(result)

result_list
```

## Reduce() and lambda functions

The reduce()-function is useful for performing some computation on a list and unlike map() and filter() , returns a single value as a result. To use reduce(), you must import it from the functools module. 

```python
# Import reduce from functools
from functools import reduce

# Create a list of strings: stark
stark = ['robb', 'sansa', 'arya', 'brandon', 'rickon']

# Use reduce() to apply a lambda function over stark: result
result = reduce(lambda item1, item2: item1 + item2, stark)

# Print the result
print(result)
```

## Error handling

Exceptions - caught during executions

Catch exeptions with try-except code


```python
def sqrt(x):
    try:
        return x** 0.5
    except:
        print('x must be an int or float')
sqrt(4)
```

```python
sqrt('hi')
```

the functions prints out what we wanted

You might only want to catch TypeErrors and let other errors pass through. The following code will ONLY catch typeerrors. If something else goes wrong, Python will still raise the correct exception. This is consedered best practice. 

```python
def sqrt(x):
    try:
        return x ** 0.5
    except TypeError:
        print('x must be an int or float')

sqrt('hey')
```

More often than not, instead of merely printing an error message, we'll want to raise an error.

We can raise any exception class:

ValueError

TypeError

RuntimeError

KeyError

IndexError

ZeroDivisionError

FileNotFoundError





for example:


raise TypeError("x must be a number")

raise ValueError("x must be positive")

raise RuntimeError("Something unexpected happened")

```python
def sqrt(x):
    if x < 0:
        raise ValueError('x must be non-negative')
    # you MANUALLY throw an error because in this case
    # a neg number will make the function behave in a weird way.
    
    try:
        return x** 0.5
    except TypeError:
        #here we CATCH an errror that Python throws
        print('x must be an int or float')
```
