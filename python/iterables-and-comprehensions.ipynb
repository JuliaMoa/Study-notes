Definition of an iterable = an object with an associated iter() method

Applying iter() to an iterable creates an iterator

Iterator = an object that has an associated next() method that produces the consecutive values

```python
word = 'Da'
it = iter(word)
next(it)
```

```python
## Iterating at once with *

word = 'Data'
it = iter(word)
print(*it)
```

Iterating over dictionaries - we need to unpack the key-value pairs while applying the items()-method

```python
pythonistas = {'hugo': 'bowne-anderson', 'francis':'castro'}
for key, value in pythonistas.items():
    print(key, value)

```

## Useful functions

enumerate() function allows us to ADD a counter to any iterable

zip() function - excepts iterables and returns an iterator of tuples

### Using enumerate()

```python
avengers = ['hawkeye', 'iron man', 'thor', 'quicksilver']
e = enumerate(avengers)
print(type(e))
```

```python
e_list = list(e)
print(e_list)
```

```python
for index, value in enumerate(avengers):
    print(index, value)

# if you don't want enumerate to start indexing at 0
# start=10
```

### Using zip()

```python
names = ['barton', 'stark', 'odinson', 'maximoff']

z = zip(avengers, names)
print(type(z))
```

When we zip two iterables we get an iterator of tuples. We can turn it into a list to print out

```python
z_list = list(z)
print(z_list)
```

alternativly we could use a for loop to iterate over the zip object and print the tuples

```python
for z1, z2 in zip(avengers, names):
    print(z1, z2)
```

We could also have used the splat-operator * to print all the elements

```python
z = zip(avengers, names)
print(*z)
```

### List comprehensions

Creates lists from other lists, DataFrame columns etc

Single line of code

More efficient than using a for loop

```python
nums = [12, 8, 21, 3, 16]
new_nums = [num + 1 for num in nums]
new_nums
```

```python
result = [num for num in range(11)]
result
```

### You can use list comprehension in place of nested loops

Might be less readable - although you get used to it

```python
matrix = [[col for col in range(1,4)] for row in range(3)]
print(matrix)
```

recreate this matrix by using nested listed comprehensions. Recall that you can create one of the rows of the matrix with a single list comprehension. To create the list of lists, you simply have to supply the list comprehension as the output expression of the overall list comprehension:

[[output expression] for iterator variable in iterable]

Note that here, the output expression is itself a list comprehension.

```python
# Create a 5 x 5 matrix using a list of lists: matrix
matrix = [[col for col in range(5)] for row in range(5)]

# Print the matrix
for row in matrix:
    print(row)
```

Conditionals in comprehensions



```python
## the resulting list is the square of the values in range(10)
## under the condition that the number itself is even


[num ** 2 for num in range(10) if num % 2 == 0]
```

```python
## we can also condition it on the output expression
## for an even int we out put its sqaure, for an odd number we output 0

[num ** 2 if num % 2 == 0 else 0 for num in range(10)]
```

We can create dictionary comprehensions. 2 differences. We use curly brackets, and the key and value are seperated by a colon in the output expression

```python
pos_neg = {num: -num for num in range(9)}
print(pos_neg)
```

```python
# Create a list of strings: fellowship
fellowship = ['frodo', 'samwise', 'merry', 'aragorn', 'legolas', 'boromir', 'gimli']

# Create dict comprehension: new_fellowship
new_fellowship = {member : len(member) for member in fellowship}

# Print the new dictionary
print(new_fellowship)
```

## Generator functions

produces generator objects

defined like a regular function - def

BUT yields a sequence of values instead of returning a single value. Generates a value with yield keyword

```python
def num_sequences(n):
    """Generate values from 0 to n."""
    i = 0
    while i < n:
        yield i
        i += 1

result = num_sequences(5)

for item in result:
    print(item)
```
