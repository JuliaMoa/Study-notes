## Working with durations

Much like dates, datetimes have a kind of arithmetic; we can compare them, subtract them, and add intervals to them

```python
start = datetime(2017, 10, 8, 23, 46, 47)
end = datetime(2017, 10, 9, 0, 10, 57)

# subtract datetimes to create a timedelta - the elapsed time 
duration = end - start

print(duration.total_seconds())
```

## Timezones

before or after UTC (standard time)

```python
from datetime import datetime, timedelta, timezone

# creating a timezone object, which accepts a timedelta that
# explains how to translate your datetime into UTC
ET = timezone(timedelta(hours=-5))

# timezone-aware datetime
dt = datetime(2017, 12, 30, 15, 9, 3, tzinfo= ET)
print(dt)
```

this is american eastern standard time

making datetime aware of its timezone means you can ask new questions - like what would the time have been in a different timetime?

```python
# india standard time zone
IST = timezone(timedelta(hours=5, minutes=30))

# convert to IST
print(dt.astimezone(IST))
```

## Adjusting timezone vs changing tzinfo

important difference

```python
print(dt.replace(tzinfo=timezone.utc))

# the clock stays the same but the utc has shifted
```

Using .replace() to change the timezone leaves the date and time the same.

Wheras if you use the .astimezone() method you shift the date and time to match the new timezone.

```python
print(dt.astimezone(timezone.utc))

# we change both the UTC offset and the clock itself
```

A datetime is 'timezone aware' when it has its itzinfo set. Otherwise its 'timezone naive'. Setting a timezone tells a datetime how to align itself to UTC, the universal time standard.

## Starting Daylight Saving time



When the clocks move forward in the spring - the clock 'springs' forward

And the utc offset is going to change

```python
spring_ahead_159am = datetime(2017, 3, 12, 1, 59, 59)
spring_ahead_159am.isoformat()
```

```python
spring_ahead_3am = datetime(2017, 3, 12, 3, 0, 0)
spring_ahead_3am.isoformat()
```

how much time has elapsed?

To fix problems with comparing datetimes we start by creating timezone objects.

```python
EST = timezone(timedelta(hours=-5))
EDT = timezone(timedelta(hours=-4))

spring_ahead_159am = spring_ahead_159am.replace(tzinfo= EST)
spring_ahead_159am.isoformat()
```

```python
spring_ahead_3am = spring_ahead_3am.replace(tzinfo = EDT)
spring_ahead_3am.isoformat()
```

```python
(spring_ahead_3am - spring_ahead_159am).seconds
```

## Ending Daylight Saving Time

a wrinkle occurs

in order to make sense of the situation we need to map everything back to UTC

```python
eastern = tz.gettz('US/Eastern')

# 2017-11-05 01:00:00
first_1am = datetime(2017, 11, 5, 1, 0, 0, tzinfo=eastern)
```

Local time. Let's check and see if this time is ambiguous, meaning we need to tell it apart somehow.

```python
tz.datetime_ambiguous(first_1am)
```

Yes, this is a time that could occur at two different UTC moments in this timezone. 

Because the UTC offset change, and when clocks are set back in autumn, the local clock repeats one houe. During that repeated hour, the same wall-clock time corresponds two different UTC instants.

Now we create a second datetime, with the same date and time. This time we call tz.enfold(), which takes the argument of the datetime we want to mark.

```python
second_1am = datetime(2017, 11, 5, 1, 0, 0, tzinfo=eastern)
second_1am = tz.enfold(second_1am)
```

enfold says, this datetime belongs to the 'second' time the wall clock struck 1 AM this day, and not the first.

Python doesn't take it into account when doing datetime arithmetics. Enfold is just a placeholders and it is up to further parts of the program to pay attention to enfold and do something with it. 

We need to convert to UTC, which is unambiguos.

```python
first_1am = first_1am.astimezone(tz.UTC)
second_1am = second_1am.astimezone(tz.UTC)

(second_1am - first_1am).total_seconds()
```

After switching to UTC python can now tell us that these 1am's are an hour apart.
