## Storytelling

The best results have no impact without proper presentation.

One of the main challenges is to convince change-averse stakeholders that some things should be done differently, especially if these stakeholders are non-technical and do not feel comfortable with math or statistics.

Why is storytellning important?

- Anecdotes = Imagination

- Stories = memorable

- Add value (provide contect)

- Capture audience's attention

- Facilitate decision-making

- Drive change


Data storytelling

- 3 minutes story: 
    - what would you say in 3 minutes
- Big idea:
    - Unique points of view
    - One sentence

--> clear and consice

#### Data - Narrative - Visuals 

These parts make the storytellning insightful, explanatory and consice.

Data 
- Results (e.g predictions) and findings (e.g. data analysis)
- Relevant 
- Accurate and reliable
- Actionable insights

Narrative
- Compelling and easy to understand
- Prioritize essential points
- Avoid disconnected facts
- Central insights
- Understand background and audience
- Clarify facts to that audience
- Linear sequence

Visuals
- Graphs should be:
    - simple
    - engaging
    - not misleading
